```python
import pandas as pd
import sqlite3
```

```python
orders = pd.read_csv("orders.csv")

print(orders.head())
```

```python
users = pd.read_json("users.json")

print(users.head())
```

```python
conn = sqlite3.connect("restaurants.db")

with open("restaurants.sql", "r", encoding="utf-8") as f:
    sql_script = f.read()

conn.executescript(sql_script)
```

```python
restaurants = pd.read_sql_query("SELECT * FROM restaurants", conn)

print(restaurants.head())
```

```python
orders_users = pd.merge(
    orders,
    users,
    on="user_id",
    how="left"
)
```

```python
final_df = pd.merge(
    orders_users,
    restaurants,
    on="restaurant_id",
    how="left"
)
```

```python
final_df.to_csv(
    "final_dataset.csv",
    index=False
)
```

```python
df = pd.read_csv("final_food_delivery_dataset.csv")
df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
df[df['membership'] == 'Gold'] \
    .groupby('city')['total_amount'] \
    .sum() \
    .sort_values(ascending=False) \
    .head(1)
```

```python
df.groupby('cuisine')['total_amount'] \
  .mean() \
  .sort_values(ascending=False) \
  .head(1)
```

```python
df.groupby('user_id')['total_amount'] \
  .sum() \
  .reset_index() \
  .query("total_amount > 1000")['user_id'] \
  .nunique()
```

```python
def rating_range(x):
    if 3.0 <= x <= 3.5:
        return "3.0 – 3.5"
    elif 3.6 <= x <= 4.0:
        return "3.6 – 4.0"
    elif 4.1 <= x <= 4.5:
        return "4.1 – 4.5"
    elif 4.6 <= x <= 5.0:
        return "4.6 – 5.0"
    else:
        return None

df['rating_range'] = df['rating'].apply(rating_range)

df.groupby('rating_range')['total_amount'] \
  .sum() \
  .sort_values(ascending=False) \
  .head(1)
```

```python
df[df['membership'] == 'Gold'] \
    .groupby('city')['total_amount'] \
    .mean() \
    .sort_values(ascending=False) \
    .head(1)
```

```python
df.groupby('cuisine').agg(
    restaurant_count=('restaurant_id', 'nunique'),
    total_revenue=('total_amount', 'sum')
).sort_values(
    by=['restaurant_count', 'total_revenue'],
    ascending=[True, False]
)
```

```python
round(
    (df['membership'] == 'Gold').sum() / len(df) * 100
)
```

```python
temp = df.groupby('restaurant_name_x').agg(
    total_orders=('order_id', 'count'),
    avg_order_value=('total_amount', 'mean')
)

temp[temp['total_orders'] < 20] \
    .sort_values(by='avg_order_value', ascending=False) \
    .head(5)
```
